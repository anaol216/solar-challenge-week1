```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import f_oneway, kruskal
```

```python
benin = pd.read_csv('data/benin_clean.csv')
sierra_leone = pd.read_csv('data/sierra_leone_clean.csv')
togo = pd.read_csv('data/togo_clean.csv')

benin['Country'] = 'Benin'
sierra_leone['Country'] = 'Sierra Leone'
togo['Country'] = 'Togo'

df = pd.concat([benin, sierra_leone, togo], ignore_index=True)
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Country', y='GHI', palette='Set2')
plt.title('GHI Distribution by Country')
plt.ylabel('GHI (kWh/m²/day)')
plt.xlabel('')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Country', y='DNI', palette='Set2')
plt.title('DNI Distribution by Country')
plt.ylabel('DNI (kWh/m²/day)')
plt.xlabel('')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Country', y='DHI', palette='Set2')
plt.title('DHI Distribution by Country')
plt.ylabel('DHI (kWh/m²/day)')
plt.xlabel('')
plt.tight_layout()
plt.show()
```

```python
metrics = ['GHI', 'DNI', 'DHI']
summary_table = df.groupby('Country')[metrics].agg(['mean', 'median', 'std']).round(2)
summary_table
```

```python
ghi_benin = benin['GHI']
ghi_sierra = sierra_leone['GHI']
ghi_togo = togo['GHI']

# Use ANOVA if data is roughly normally distributed, else use Kruskal–Wallis
f_stat, p_anova = f_oneway(ghi_benin, ghi_sierra, ghi_togo)
h_stat, p_kruskal = kruskal(ghi_benin, ghi_sierra, ghi_togo)

print(f'ANOVA p-value: {p_anova:.4f}')
print(f'Kruskal-Wallis p-value: {p_kruskal:.4f}')
```

```python
avg_ghi = df.groupby('Country')['GHI'].mean().sort_values(ascending=False)

plt.figure(figsize=(6, 4))
sns.barplot(x=avg_ghi.values, y=avg_ghi.index, palette='Set3')
plt.xlabel('Average GHI (kWh/m²/day)')
plt.title('Average GHI by Country')
plt.tight_layout()
plt.show()
```
